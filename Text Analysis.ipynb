```python
import os
import re
import string


import nltk
from nltk.tokenize import word_tokenize
from nltk import FreqDist
nltk.download('punkt_tab')

import spacy
nlp = spacy.load("en_core_web_sm")
from spacy import displacy

import pandas as pd
```

```python
def get_text_info(text):
    tokens = nltk.word_tokenize(text)
    n_tokens = len(tokens)
    n_types = len(set(tokens))
    diversity = n_types / n_tokens

    print("Length: %2d" % (n_tokens))
    print("Lexical Diversity: %3f" % (diversity))


def top_ten_frequent_words(text):
    words = word_tokenize(text.lower())
    cleaned_words = []
    for word in words:
        if word not in string.punctuation and word != '“' and word != '”' and word != '\'s':
          cleaned_words.append(word)
    freqdist = FreqDist(cleaned_words)
    top_ten_words = freqdist.most_common(10)
    print(top_ten_words)

def get_named_entities(text):
    doc = nlp(text)
    entity_list = []
    for ent in doc.ents:
        entity_list.append((ent.text, ent.label_))

    df_ents = pd.DataFrame(entity_list)
    df_ents.columns = ['Entity', 'Label']
    print(df_ents)


def process_dir(path):
    file_info = {}

    for filename in os.listdir(path):
        if filename.endswith(".txt"):
            file_path = os.path.join(path, filename)
            with open(file_path, 'r', encoding="utf-8") as f:
                text = f.read()
                print("\n",filename,"\n")
                get_text_info(text)
                top_ten_frequent_words(text)
                get_named_entities(text)
```

```python
path = './data'

files_in_dir_info = process_dir(path)
```
